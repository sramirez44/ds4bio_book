## Control flow

###  If statements

+ Python has if statements like most programming languages
+ Notice the indentation and the `:`

````python
# do this if you'd like to prompt for an input
x = input("are you mean (y/n)? > ")
# Let's just assume the user input 'n' and just assign x='n'
# x = 'n'
if x == 'y': 
 print("Slytherine!")
elif x == 'n':
 print("Gryffindor")
else:
 print("please enter y or n exactly")
````

### White space

+ White space is more meaningful in python than other languages
+ This is actually useful, as it forces good indentation practices

```
## Some more about white space
if statementA:
  statementB   # Executed if statementA is True
statementC     # Executed regardless since it's not indented
```

### General if statment structure


```
if statement1 :
 ...
elif statement2 :
 ...
else 
 ...
```
Here's an example (note this is just equal to the statement `(a < 0) - (a > 0)`

````python
a = 5

if a < 0 :
  a = -1
elif a > 0 :
  a = 1
else :
  a = 0

print(a)
````

## for and while loops

### looping in python

+ Python has for and while loops 
+ for loops will iterate over most objects for which it makes sense
+ Don't forget the indentation or the `:`


````python
for i in "test":
    print(i)
````

````python
for i in [[1, 2], [3, 4]]:
    print(i)
````

### range
+ range is a function that creates an iterable list of integers
+ `range(4)` is not itself a list, but `list(range(4))` is

````python
print(type(range(4)))
print(type(list(range(4))))
````

````python
for i in range(4) :
 print(i)
````

### While loop example

````python
x = 4
while x > 0 :
 x = x - 1
 print(x)
````

````python
## Note for loops can iterate over list-like structures.
for w in 'word':
    print(w)
````

````python
for w in ['word1','word2','word3']:
    for w2 in w :
       print(w2)
````

````python
for w in [['word1','word2','word3'],['word4']]:
    print(w)
````

````python
## the range function is useful for creating a structure to loop over. It creates a data type that
## can be iterated over, but isn't itself a list. If you want a list out of it, you have to convert it

a = range(3)
print(a)
print(list(a)) # conversion
````

````python
# list comprehesion
[x**2  + 4 for x in range(4)] # convenient way to create a list from an output of a for loop
````

````python
a = list(range(5))
b = list(range(6,11))

[a[i] + b[i] for i in range(len(a))] # looping over the length of a (indeces) and then taking the ith
# element of a and the ith element of b and adding them togeth
````
